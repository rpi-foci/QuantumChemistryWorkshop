```python
from qiskit.circuit import QuantumCircuit
from qiskit.quantum_info import SparsePauliOp
from qiskit.primitives import StatevectorEstimator

# Create a simple two-qubit circuit
qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0, 1)

# Write your observable in terms of Pauli strings
H = SparsePauliOp(["ZZ", "XI"], coeffs=[1.0, 0.5])

# Estimate the expectation value of this observable
estimator = StatevectorEstimator()
result = estimator.run([(qc, H)]).result()
print("Expectation Value:", result[0].data.evs)
```
