```python
from qiskit import QuantumCircuit

 #Create a quantum circuit with two qubits
qc = QuantumCircuit(2)
# Apply H gate to qubit 0
qc.h(0)
# Apply CNOT control 0, target 1
qc.cx(0,1)
# Apply CZ control 0, target 1
qc.cz(0,1)
# Measure qubits                          
qc.measure_all()
# Draw circuit
qc.draw()
```

```python
# Draw it nicer
qc.draw(output='mpl')
```
